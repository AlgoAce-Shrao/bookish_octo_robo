414.Third Maximum Number
O[n]
Best solution

```python
class Solution {
    public int thirdMax(int[] nums) {
        Set<Integer> s=new HashSet<>(); //making numbers unique
        for(int n:nums) s.add(n);
        if(s.size()<3) return Collections.max(s); //if less then 3 elements..return maximum of them
        else{
            s.remove(Collections.max(s));//remove first max
            s.remove(Collections.max(s));//remove seconc max
        }
        return Collections.max(s);//print third max
    }
}
```
