2089.Find target indices after sorting array
My solution...-->O[N^2]

```python
class Solution {
    public List<Integer> targetIndices(int[] nums, int target) {
        int n=nums.length;
        ArrayList <Integer> ls=new ArrayList<>();
        for(int i=0;i<=(n-1);i++){
            int min=i;
            for(int j=i;j<=(n-1);j++){
                if(nums[j]<nums[min]) min=j;
            }
            int temp;
            temp=nums[i];
            nums[i]=nums[min];
            nums[min]=temp;
        }
        for(int i=0;i<n;i++){
            if(nums[i]==target){
                ls.add(i);
            }
        }
        Collections.sort(ls);
        return ls;
    }
}
```
